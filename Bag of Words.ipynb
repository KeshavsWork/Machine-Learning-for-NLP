```python
import pandas as pd
df = pd.read_csv("Spam-classifier Master/spam.csv", encoding="latin1")

# Keep only the first two columns: v1 = label, v2 = message
df = df[['v1', 'v2']]

# Rename columns for clarity
df = df.rename(columns={'v1': 'label', 'v2': 'message'})
```

```python
df
```

```python
# # Data Cleaning and Preprocessing
import re
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
ps = PorterStemmer()
```

```python
#Using SnowballStemmer
from nltk.stem import SnowballStemmer
snowballstemmer=SnowballStemmer('english')
```

```python
# Using Lemmatization
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
```

```python
lemma=WordNetLemmatizer()
```

```python
corpus=[]
for i in range(0,len(df)):
    review=re.sub('[^a-zA-Z]',' ',df['message'][i])
    review=review.lower()
    review=review.split()
    # review=[ps.stem(word) for word in review if word not in stopwords.words('english')] 
    # review=[snowballstemmer.stem(word) for word in review if word not in stopwords.words('english')]
    review=[lemma.lemmatize(word,pos='v') for word in review if word not in stopwords.words('english')]
    review=' '.join(review)
    corpus.append(review)
```

# Create Bag of Words

```python
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer(max_features=2500)
X=cv.fit_transform(corpus).toarray()
```

```python
print(X[:5])
```

```python
cv=CountVectorizer(max_features=100,binary=True)
X=cv.fit_transform(corpus).toarray()
print(X)
```
