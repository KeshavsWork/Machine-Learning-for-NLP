# 📌 Named Entity Recognition (NER) Tags with Examples

| **NER Tag**   | **Full Form**                                       | **Example Sentence**                       | **Entity Extracted** |
| ------------- | --------------------------------------------------- | ------------------------------------------ | -------------------- |
| PERSON        | People, including fictional                         | *"Barack Obama was the president."*        | Barack Obama         |
| NORP          | Nationalities, religious groups, political groups   | *"The Americans celebrated Thanksgiving."* | Americans            |
| FAC           | Facilities (buildings, airports, highways, bridges) | *"Heathrow Airport is in London."*         | Heathrow Airport     |
| ORG           | Organizations (companies, agencies, institutions)   | *"Google was founded in 1998."*            | Google               |
| GPE           | Geo-Political Entity (countries, cities, states)    | *"I live in India."*                       | India                |
| LOC           | Non-GPE locations (mountains, bodies of water)      | *"Mount Everest is the tallest mountain."* | Mount Everest        |
| PRODUCT       | Objects, vehicles, food, etc.                       | *"She bought an iPhone."*                  | iPhone               |
| EVENT         | Named events (wars, sports, disasters)              | *"The Olympics were held in Tokyo."*       | Olympics             |
| WORK\_OF\_ART | Titles of books, songs, works                       | *"I love reading Harry Potter."*           | Harry Potter         |
| LAW           | Named laws, treaties, regulations                   | *"The Indian Penal Code is important."*    | Indian Penal Code    |
| LANGUAGE      | Named languages                                     | *"He speaks French."*                      | French               |
| DATE          | Absolute/relative dates or periods                  | *"He was born on July 4th, 1990."*         | July 4th, 1990       |
| TIME          | Specific times                                      | *"The meeting is at 5 PM."*                | 5 PM                 |
| PERCENT       | Percentage (with %)                                 | *"He scored 95% on the exam."*             | 95%                  |
| MONEY         | Monetary values                                     | *"The phone costs \$699."*                 | \$699                |
| QUANTITY      | Measurements (weight, distance, height)             | *"The bridge is 5 km long."*               | 5 km                 |
| ORDINAL       | First, second, etc.                                 | *"He came in 1st place."*                  | 1st                  |
| CARDINAL      | Numerals not referring to anything else             | *"There are 300 students in the school."*  | 300                  |


```python
sentence="Barack Obama was born in Hawaii and served as the 44th President of the United States."
```

```python
import nltk
words=nltk.word_tokenize(sentence)
```

```python
words
```

```python
tag_elements=nltk.pos_tag(words)
print(tag_elements)
```
